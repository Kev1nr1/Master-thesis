```python
from collections import Counter

# List of all words in exploited and not exploited CVEs
exploited_words = [word for words in df_exploited["description_cleaned"] for word in words]
not_exploited_words = [word for words in df_not_exploited["description_cleaned"] for word in words]

# Count their word frequencies
exploited_word_counts = Counter(exploited_words)
not_exploited_word_counts = Counter(not_exploited_words)

exploited_total = len(exploited_words)
not_exploited_total = len(not_exploited_words)

# Normalize the counted CVEs and calculate the difference
normalized_exploited = {word: count / exploited_total for word, count in exploited_word_counts.items()}
normalized_not_exploited = {word: count / not_exploited_total for word, count in not_exploited_word_counts.items()}
word_diff = {word: normalized_exploited.get(word, 0) - normalized_not_exploited.get(word, 0) for word in set(normalized_exploited) | set(normalized_not_exploited)}

# Print top 20
print("Top 20 words more common in exploited CVEs:")
for word, diff in sorted(word_diff.items(), key=lambda item: item[1], reverse=True)[:20]:
    print(f"{word}: {diff:.4f}")
    

    
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(df["description_cleaned"].apply(lambda x: ' '.join(x)))

tfidf_feature_names = vectorizer.get_feature_names_out()

def top_words_tfidf(filter, title):
    filtered_tfidf_matrix = tfidf_matrix[filter]

    avg_scores_high = filtered_tfidf_matrix.mean(axis=0).A1
    top_words_df = pd.DataFrame({
        'word': tfidf_feature_names,
        'avg_tfidf_score': avg_scores_high
    })

    top_words_df = top_words_df.sort_values(by='avg_tfidf_score', ascending=False)
    print(title)
    print(top_words_df.head(15))
    
top_words_tfidf((df['combined_score'] >= 6).values, "Top Words for CVEs with Combined Score >= 6")
print()
top_words_tfidf((df['combined_score'] < 6).values, "Top Words for CVEs with Combined Score < 6")
```

```python
# Sourced from:
# https://scikit-learn.org/stable/auto_examples/applications/plot_topics_extraction_with_nmf_lda.html
import matplotlib.pyplot as plt


def plot_top_words(model, feature_names, n_top_words, title):
    fig, axes = plt.subplots(2, 5, figsize=(30, 15), sharex=True)
    axes = axes.flatten()
    for topic_idx, topic in enumerate(model.components_):
        top_features_ind = topic.argsort()[-n_top_words:]
        top_features = feature_names[top_features_ind]
        weights = topic[top_features_ind]

        ax = axes[topic_idx]
        ax.barh(top_features, weights, height=0.7)
        ax.set_title(f"Topic {topic_idx +1}", fontdict={"fontsize": 30})
        ax.tick_params(axis="both", which="major", labelsize=20)
        for i in "top right left".split():
            ax.spines[i].set_visible(False)
        fig.suptitle(title, fontsize=40)

    plt.subplots_adjust(top=0.90, bottom=0.05, wspace=0.90, hspace=0.3)
    plt.show()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
from tqdm import tqdm

vectorizer_total = CountVectorizer(min_df=5, max_df=0.95)
tf_total = vectorizer_total.fit_transform(df["description_cleaned"].apply(lambda x: ' '.join(x)))

lda_total = LatentDirichletAllocation(n_components=10, random_state=1, learning_method='online', learning_offset=50.0, n_jobs=-1)

n_iter = 10
with tqdm(total=n_iter, desc="LDA Training") as pbar:
    for i in range(n_iter):
        lda_total.partial_fit(tf_total)
        pbar.update(1)
tf_feature_names_total = vectorizer_total.get_feature_names_out()
plot_top_words(lda_total, tf_feature_names_total, 10, "Topics in LDA model")
```

```python

def get_top_words_distinct(model, feature_names, n_top=10):
    seen = set()
    words = []
    
    for i, topic in enumerate(model.components_):
        sorted_idx = topic.argsort()[::-1]
        count = 0
        for idx in sorted_idx:
            word = feature_names[idx]
            if word not in seen:
                seen.add(word)
                words.append(word)
                count += 1
            if count == n_top:
                break
    
    return words

distinct_words_smaller = get_top_words_distinct(lda_combined_smaller, tf_feature_names_combined_smaller, n_top=10)
distinct_words_bigger = get_top_words_distinct(lda_combined_bigger, tf_feature_names_combined_bigger, n_top=10)

set_smaller = set(distinct_words_smaller)
set_bigger = set(distinct_words_bigger)

only_in_smaller = set_smaller - set_bigger
only_in_bigger = set_bigger - set_smaller

print("Only in SMALLER:")
for word in only_in_smaller:
    print(f"  {word}")

print("\nOnly in BIGGER:")
for word in only_in_bigger:
    print(f"  {word}")
```
