Loading in the fetched data

```python
# Importing 10 year dataset
import pandas as pd

df = pd.read_csv('datasets/vulnerabilities_10_years.csv')
df
```

Initial cleaning

```python
# Check for missing values in the dataset
df.isna().sum()
```

```python
# Data cleaning:
initial_count = len(df)

# Removing invalid -1 base scores
before = len(df)
df = df[df['baseScore'] != -1]
print(f"Removed invalid -1 base scores: {before - len(df)}")

# Removing rows with NaN baseScoreVector
before = len(df)
df = df[df['baseScoreVector'].notna()]
print(f"Removed NaN baseScoreVector: {before - len(df)}")

# Removing rejected/duplicate CVEs
before = len(df)
df = df[~df['description'].str.startswith('** REJECT **') & 
        ~df['description'].str.startswith('DO NOT USE THIS CANDIDATE') & 
        ~df['description'].str.startswith('Rejected reason:') &
        ~df['description'].str.startswith("This CVE ID has been rejected") &
        ~df['description'].str.startswith("This CVE is a duplicate of another CVE")]
print(f"Removed rejected/duplicate CVEs: {before - len(df)}")

# Check for leftover combination of base score 0 and empty enisaIdProduct and enisaIdVendor
incorrect_rows = (
    df["baseScore"].isin([0, 0.0]).fillna(True) &
    df["enisaIdProduct"].apply(lambda x: x == [] or x is None) &
    df["enisaIdVendor"].apply(lambda x: x == [] or x is None)
)
print(f"Remaining base score 0 with empty product/vendor: {incorrect_rows.sum()}")

print(f"Total removed: {initial_count - len(df)} ({initial_count} -> {len(df)})")
```

```python
# Check for missing values in the dataset
df.isna().sum()
```

```python
# Manually go over the remaining 69 references and check if they are valid
left_over = df[df['references'].isna()==True]
```

```python
# Leftover references were invalid as well and will be removed from the dataset
df = df[df['references'].isna()==False]
```

Description cleaning

```python
import nltk
from nltk.corpus import stopwords
nltk.download('wordnet')  
nltk.download('punkt')     
nltk.download('stopwords')   
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from tqdm import tqdm

# Process bar + lemmatization
tqdm.pandas()
lemmatizer = WordNetLemmatizer()

# Remove special characters, lowercase and lemmatize
df["description_cleaned"] = df["description"].str.replace(r'[^\w\s]', '', regex=True).str.lower()
df["description_cleaned"] = df["description_cleaned"].progress_apply(lambda x: [lemmatizer.lemmatize(word) for word in word_tokenize(x)])

# Remove stop words and generate word cloud
stop_words = set(stopwords.words('english'))
df["description_cleaned"] = df["description_cleaned"].progress_apply(lambda x: [word for word in x if word not in stop_words])
```

Product and vendor filtering

```python
import ast
import pandas as pd

def extract_name(value, key):
    if isinstance(value, str):
        try:
            value = ast.literal_eval(value)
        except (ValueError, SyntaxError):
            return None

    if isinstance(value, list) and value:
        return value[0].get(key, {}).get("name")

    return None

# Update original dataframe
df["product_name"] = df["enisaIdProduct"].apply(
    lambda x: extract_name(x, "product")
)

df["vendor_name"] = df["enisaIdVendor"].apply(
    lambda x: extract_name(x, "vendor")
)

# Create separate dataframe with unique combinations
result_df = (
    df[["product_name", "vendor_name"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

result_df = result_df.dropna()
result_df = result_df[result_df["product_name"] != "n/a"]
result_df = result_df[result_df["vendor_name"] != "n/a"]
```

Saving to CSV file

```python
result_df.to_csv("datasets/product_vendor_combinations.csv", index=False)
```
