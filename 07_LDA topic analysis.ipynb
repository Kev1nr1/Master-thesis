```python
import pandas as pd

df = pd.read_csv('datasets/labeled_vulnerabilities.csv')
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
from gensim.models.coherencemodel import CoherenceModel
from gensim.corpora import Dictionary
import numpy as np
import matplotlib.pyplot as plt

# ── Data prep ────────────────────────────────────────────────────────
tokenized_docs = df["description_cleaned"].tolist()  # lists of tokens
texts = [" ".join(doc) for doc in tokenized_docs]    # joined strings for vectorizer

# ── 1. Vectorizer ────────────────────────────────────────────────────
vectorizer = CountVectorizer(
    max_df=0.5,
    min_df=10,
    max_features=50000,
    ngram_range=(1, 2),
    token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z0-9_\-]{2,}\b",
    stop_words='english',
)
dtm = vectorizer.fit_transform(texts)
feature_names = vectorizer.get_feature_names_out()

# ── 2. Gensim dictionary (for coherence only) ────────────────────────
gensim_dict = Dictionary(tokenized_docs)

def get_top_words_per_topic(model, feature_names, n_words=10):
    topics = []
    for topic in model.components_:
        top_indices = topic.argsort()[:-n_words - 1:-1]
        topics.append([feature_names[i] for i in top_indices])
    return topics

# ── 3. k search via coherence ────────────────────────────────────────
k_range = [10, 15, 20, 25, 30, 35, 40]
scores = []

for k in k_range:
    lda = LatentDirichletAllocation(
        n_components=k,
        doc_topic_prior=0.1,    # low alpha = sparse, distinctive topic assignments
        topic_word_prior=None,  # auto beta
        max_iter=20,
        learning_method='batch',
        random_state=42,
    )
    lda.fit(dtm)

    topics = get_top_words_per_topic(lda, feature_names, n_words=10)
    cm = CoherenceModel(
        topics=topics,
        texts=tokenized_docs,
        dictionary=gensim_dict,
        coherence='c_v',
    )
    coherence = cm.get_coherence()
    scores.append((k, coherence))
    print(f"k={k}: coherence={coherence:.4f}")

best_k = max(scores, key=lambda x: x[1])[0]
print(f"\nBest k by coherence: {best_k}")

# ── 4. Plot ──────────────────────────────────────────────────────────
ks, coherences = zip(*scores)
plt.figure(figsize=(7, 4))
plt.plot(ks, coherences, marker='o', color='darkorange')
plt.xlabel("Number of Topics (k)")
plt.ylabel("Coherence score (c_v)")
plt.title("LDA Coherence vs. k")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ── 5. Final model ───────────────────────────────────────────────────
lda_final = LatentDirichletAllocation(
    n_components=best_k,
    doc_topic_prior=0.1,
    topic_word_prior=None,
    max_iter=20,
    learning_method='batch',
    random_state=42,
)
lda_final.fit(dtm)

# ── 6. Inspect topics ────────────────────────────────────────────────
def print_topics(model, feature_names, n_words=10):
    for i, topic in enumerate(model.components_):
        top_words = [feature_names[j] for j in topic.argsort()[:-n_words - 1:-1]]
        print(f"Topic {i:02d}: {', '.join(top_words)}")

print_topics(lda_final, feature_names)

# ── 7. Document-topic assignment ─────────────────────────────────────
doc_topics = lda_final.transform(dtm)
df["dominant_topic"] = doc_topics.argmax(axis=1)
```
