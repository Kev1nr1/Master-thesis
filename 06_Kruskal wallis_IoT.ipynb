```python
import pandas as pd

df = pd.read_csv('datasets/labeled_vulnerabilities_IoT.csv', index_col=0)
df = df[df['product_label'] != 'Other category'].copy()
df
```

```python
from scipy.stats import kruskal

groups_epss = [
    group["epss"].dropna().values
    for _, group in df.groupby("product_label")
]

stat, p = kruskal(*groups_epss)

print(f"H = {stat:.3f}")
print(f"p = {p:.5f}")

groups_cvss = [
    group["baseScore"].dropna().values
    for _, group in df.groupby("product_label")
]

stat, p = kruskal(*groups_cvss)

print(f"H = {stat:.3f}")
print(f"p = {p:.5f}")
```

```python
n = len(df)
k = df["product_label"].nunique()

def epsilon_squared(H, n):
    return H / (n - 1)

eps_epss = epsilon_squared(7250.029, n)
eps_cvss = epsilon_squared(7954.899, n)

print(f"EPSS effect size (epsilon^2): {eps_epss:.4f}")
print(f"CVSS effect size (epsilon^2): {eps_cvss:.4f}")
```

```python
group_sizes = df.groupby("product_label")["epss"].count().sort_values()
print("15 smallest groups:")
print(group_sizes.sort_values().head(15))

print("\n15 largest groups:")
print(group_sizes.sort_values(ascending=False).head(15))
```

```python
import scikit_posthocs as sp

import numpy as np

dunn_epss = sp.posthoc_dunn(df, val_col="epss", group_col="product_label", p_adjust="holm")

# Flatten the matrix into pairwise comparisons
mask = np.triu(np.ones(dunn_epss.shape), k=1).astype(bool)
results_flat = dunn_epss.where(mask).stack().reset_index()
results_flat.columns = ["group1", "group2", "p_adj"]
results_flat = results_flat.sort_values("p_adj").reset_index(drop=True)

# Significant pairs only
sig_pairs = results_flat[results_flat["p_adj"] < 0.05]


import pandas as pd

appearance_counts = pd.concat([sig_pairs["group1"], sig_pairs["group2"]]).value_counts()
print(appearance_counts)
```

```python
dunn_cvss = sp.posthoc_dunn(df, val_col="baseScore", group_col="product_label", p_adjust="holm")

mask = np.triu(np.ones(dunn_cvss.shape), k=1).astype(bool)
cvss_flat = dunn_cvss.where(mask).stack().reset_index()
cvss_flat.columns = ["group1", "group2", "p_adj"]
cvss_flat = cvss_flat.sort_values("p_adj").reset_index(drop=True)

sig_pairs_cvss = cvss_flat[cvss_flat["p_adj"] < 0.05]
appearance_counts_cvss = pd.concat([sig_pairs_cvss["group1"], sig_pairs_cvss["group2"]]).value_counts()
print(appearance_counts_cvss)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import textwrap

# Wrap long labels
def wrap_label(label, width=40):
    return "\n".join(textwrap.wrap(label, width))

df["label_wrapped"] = df["product_label"].apply(wrap_label)

# Sort by median EPSS for visual clarity
order = (
    df.groupby("label_wrapped")["epss"]
    .median()
    .sort_values(ascending=True)
    .index.tolist()
)

n_cats = df["label_wrapped"].nunique()
fig, ax = plt.subplots(figsize=(14, max(8, n_cats * 0.6)))

sns.boxenplot(
    data=df,
    x="epss",
    y="label_wrapped",
    order=order,
    orient="h",
    color="steelblue",
    linewidth=0.8,
    ax=ax,
)

ax.set_title("EPSS Score Distribution by Product Category", fontsize=14, pad=14)
ax.set_xlabel("EPSS Score (probability of exploitation within 30 days)", fontsize=11)
ax.set_ylabel("")  # labels are self-explanatory
ax.set_xlim(left=0)
ax.tick_params(axis="y", labelsize=9)
ax.tick_params(axis="x", labelsize=10)
ax.xaxis.grid(True, linestyle="--", alpha=0.5)
ax.set_axisbelow(True)

sns.despine(left=True)
plt.tight_layout()
plt.savefig("epss_boxenplot.png", dpi=150, bbox_inches="tight")
plt.show()
```
