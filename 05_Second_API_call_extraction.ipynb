```python
import json
import pandas as pd

def extract_json_arrays(filepath):
    with open(filepath, "r", encoding="utf-8") as f:
        text = f.read()

    arrays = []
    start = None
    depth = 0
    in_string = False
    escaped = False

    for i, char in enumerate(text):

        if char == "\\" and in_string:
            escaped = not escaped
            continue

        if char == '"' and not escaped:
            in_string = not in_string

        escaped = False

        if in_string:
            continue

        if char == "[":
            if depth == 0:
                start = i
            depth += 1

        elif char == "]":
            depth -= 1

            if depth == 0 and start is not None:
                arrays.append((start, text[start:i + 1]))
                start = None

    return arrays


records = []
failed_blocks = []

for position, block in extract_json_arrays("datasets/secondary ai run.txt"):
    try:
        data = json.loads(block)

        if isinstance(data, list):
            records.extend(data)

    except json.JSONDecodeError as e:
        failed_blocks.append({
            "position": position,
            "error": str(e),
            "content": block
        })

df_final = pd.DataFrame(records)

# Save failed blocks for manual review
if failed_blocks:
    with open("datasets/failed_blocks_2.txt", "w", encoding="utf-8") as f:
        for i, failure in enumerate(failed_blocks, 1):
            f.write(f"===== FAILED BLOCK {i} =====\n")
            f.write(f"Position: {failure['position']}\n")
            f.write(f"Error: {failure['error']}\n\n")
            f.write(failure["content"])
            f.write("\n\n\n")
print(f"Loaded {len(df_final)} rows")
print(f"Found {len(failed_blocks)} failed blocks")
```

```python
# Merge original_df with df_final to add product label based on vendor and product match
df = original_df.merge(
    df_final.rename(columns={
        'Original Vendor': 'vendor_match',
        'Original Product': 'product_match',
        'Generic Category': 'product_label'
    }),
    left_on=['vendor_name', 'product_name'],
    right_on=['vendor_match', 'product_match'],
    how='left'
).drop(columns=['vendor_match', 'product_match'])

print(f"Added product_label column. Rows with product_label: {df['product_label'].notna().sum()}")
```

```python
import pandas as pd

# Load the classification mapping
classification = pd.read_csv("datasets/classification.csv")
classification.columns = ["LineNum", "Category", "Status", "MergeTarget"]

# Build a lookup dict: original category -> final category (or None if REMOVE)
def resolve(row):
    if row["Status"] == "MATCH":
        return row["Category"]
    elif row["Status"] == "MERGE":
        return row["MergeTarget"]
    else:  # REMOVE
        return None

classification["FinalCategory"] = classification.apply(resolve, axis=1)
mapping = dict(zip(classification["Category"], classification["FinalCategory"]))

# Apply mapping to your dataframe
df["product_label"] = df["product_label"].map(mapping)

# Drop rows that map to None (i.e. were REMOVE)
df_clean = df.dropna(subset=["product_label"]).copy()

# Optional: rebuild counts/percentages after merging
summary = (
    df_clean["product_label"]
    .value_counts()
    .reset_index()
)
summary.columns = ["Category", "Product Count"]
summary["Percent of Total"] = (
    summary["Product Count"] / summary["Product Count"].sum() * 100
).round(2)

summary.to_csv("datasets/label_distributions.csv", index=False)
```
