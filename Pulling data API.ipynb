```python
import requests
import csv
import time

url = "https://euvdservices.enisa.europa.eu/api/search"

params = {
    "fromDate": "2016-04-21",
    "toDate": "2026-04-21",
    "page": 0,
    "size": 100
}

OUTPUT_FILE = "vulnerabilities_10_years.csv"
total_saved = 70800
START_PAGE = 708
total_pages = 2664

# Read existing fields from the CSV header
with open(OUTPUT_FILE, "r", encoding="utf-8") as f:
    fields = next(csv.reader(f))

print(f"Resuming from page {START_PAGE}, fields: {fields}")

with open(OUTPUT_FILE, "a", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fields)

    for page in range(START_PAGE, total_pages):
        try:
            params["page"] = page
            response = requests.get(url, params=params, timeout=30)
            response.raise_for_status()
            items = response.json().get("items", [])

            for item in items:
                try:
                    writer.writerow({field: item.get(field, "") for field in fields})
                    total_saved += 1
                except Exception as e:
                    print(f"  Skipping record on page {page}: {e}")
                    continue

            file.flush()
            print(f"Page {page + 1}/{total_pages} — {total_saved} records saved")

        except Exception as e:
            print(f"Skipping page {page}: {e}")
            continue

        time.sleep(0.5)

print(f"Done. {total_saved} records saved to {OUTPUT_FILE}")
```

```python
#Hash for privacy and drop the source 
import pandas as pd
df = pd.read_csv("vulnerabilities_10_years.csv")
print(df)

import hashlib

def hash_vendor(vendor_name):
    return hashlib.md5(vendor_name.strip().lower().encode()).hexdigest()[:16]
df['assigner'] = df['assigner'].apply(hash_vendor)
df.drop(columns=['references'], inplace=True)
df.to_csv("vulnerabilities_hashed_10_years.csv", index=False)
```
